##PANDAS SERIES

**1. Creating Pandas Series**

```python
import pandas as pd
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)


print(marks)
```

**2. Indexing and Slicing**

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank 1', 'Rank 2', 'Rank 3', 'Rank 4', 'Rank 5']
)

print(marks)
print(f'')
print("1st Rank:", marks.iloc[0])
print(f'')
print(f"Top 3 Ranks:")
print(marks.loc[['Rank 1', 'Rank 2', 'Rank 3']])
print(f'')
print("\n3rd Rank:", marks.iloc[2])
print(f'')
print("\nMarks greater than 90")
print(marks[marks > 90])
```

**3. Manipulating Series**

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)


marks.loc['Rank1'] = 100

marks = marks.drop('Rank5')

cgpa = marks / 10

print(f"Updated Marks:")
print(marks)
print(f'')
print(f"CGPA:")
print(cgpa)
```

**2. Data Exploration**

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})


print(f"Transactions DataFrame:")
print(transactions)

print(f'')
print(f"First 5 rows:")
print(transactions.head())

print(f'')
print(f"Last 5 rows:")
print(transactions.tail())

print(f'')
print(f"Shape:")
print(transactions.shape)

print(f'')
print(f"Column Names:")
print(transactions.columns)
print(f'')
print(f"Data Types:")
print(transactions.dtypes)

print(f'')
print(f"ProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

print(f'')
print(f"Last 3 Columns:")
print(transactions.iloc[:, -3:])


filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print(f'')
print(f"North Region with Amount greater than 200:")
print(filtered_data)

print(f'')
print(f"Product Category Counts:")
print(transactions['ProductCategory'].value_counts())

print(f'')
print(f"Unique Regions:")
print(transactions['Region'].unique())
print(f'')
print(f"Mean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

**3. Manipulating the DataFrame**

```python
print("Initial Data:")
print(transactions)

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

transactions['Discount'] = transactions['Amount'] * 0.10
print(f'')
print("After adding Discount:")
print(transactions)


transactions = transactions[transactions['TransactionID'] != 109]
print(f'')
print("After removing TransactionID 109:")
print(transactions)

transactions = transactions.drop('Discount', axis=1)
print(f'')
print("Final DataFrame:")
print(transactions)
```
